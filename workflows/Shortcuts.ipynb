```python
import os
import time
import subprocess

# --- Configurazione ---
APP_DIR = "/tmp/comfyui/frontends/aliexpress"
APP_SCRIPT_NAME = "app.py"

# --- Vai nella cartella corretta ---
os.chdir(APP_DIR)
print("📁 Directory cambiata in:", os.getcwd())

# --- 1. Termina il Processo Esistente ---
print("🛑 1. Tentativo di terminare l'istanza precedente di Gradio...")

try:
    subprocess.run(
        f"pkill -f {APP_SCRIPT_NAME} || true",
        shell=True,
        check=True,
        stdout=subprocess.PIPE,
        stderr=subprocess.PIPE
    )
    print("✅ Istanza precedente terminata.")
except subprocess.CalledProcessError as e:
    print(f"⚠️ Errore durante la terminazione: {e.stderr.decode()}")

# --- 2. Attesa ---
time.sleep(1)

# --- 3. Avvio ---
print("🚀 3. Avvio dell'app Gradio...")

try:
    subprocess.run(
        f"python {APP_SCRIPT_NAME}",
        shell=True,
        check=True
    )
except subprocess.CalledProcessError:
    print("\n🛑 Server fermato.")
```
