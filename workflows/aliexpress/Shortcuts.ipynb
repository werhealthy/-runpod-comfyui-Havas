```python
import os
import time
import subprocess

# --- Configurazione ---
APP_SCRIPT_NAME = "app.py"

# --- 1. Termina il Processo Esistente ---
print("🛑 1. Tentativo di terminare l'istanza precedente di Gradio...")

try:
    # Usiamo 'pkill -f' per terminare il processo precedente
    subprocess.run(
        f"pkill -f {APP_SCRIPT_NAME} || true",
        shell=True,
        check=True,
        stdout=subprocess.PIPE,
        stderr=subprocess.PIPE
    )
    print("✅ Istanza precedente terminata (o non era in esecuzione).")

except subprocess.CalledProcessError as e:
    print(f"⚠️ Errore durante la terminazione: {e.stderr.decode()}")
    
# --- 2. Breve attesa (Wait) ---
WAIT_SECONDS = 1
print(f"⏳ 2. Attesa di {WAIT_SECONDS} secondo per liberare la porta...")
time.sleep(WAIT_SECONDS)

# --- 3. Riavvia l'Applicazione e Stampa l'Output ---
print("🚀 3. Avvio di Gradio (L'output apparirà qui sotto e la cella sarà bloccata)...")

try:
    # Eseguiamo il comando senza '&' per NON mandarlo in background.
    # L'output standard (stdout) e l'errore standard (stderr) del comando
    # saranno quindi reindirizzati direttamente alla cella di Jupyter.
    subprocess.run(
        f"python {APP_SCRIPT_NAME}",
        shell=True,
        check=True # Questo farà apparire l'output/log nella cella
    )
    
except subprocess.CalledProcessError as e:
    # Se il server viene interrotto (ad esempio con Ctrl+C o il pulsante Stop di Jupyter),
    # il comando subprocess.run fallisce, ma è il comportamento atteso.
    print("\n✅ Server fermato dall'utente.")
    
except Exception as e:
    print(f"\n❌ Errore durante l'avvio: {str(e)}")
```
